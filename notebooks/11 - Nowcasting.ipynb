```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 1: Load the dataset from the given URL
url = 'https://infra.datos.gob.ar/catalog/sspm/dataset/143/distribution/143.3/download/emae-valores-anuales-indice-base-2004-mensual.csv'
data = pd.read_csv(url)

# Step 2: Convert 'indice_tiempo' to datetime and filter data starting from 2017
data['indice_tiempo'] = pd.to_datetime(data['indice_tiempo'])
data_2017_onwards = data[data['indice_tiempo'] >= '2017-01-01']

# Step 3: Extract the trend and the last 4 points for linear extrapolation
last_4_trend = data_2017_onwards[['indice_tiempo', 'emae_tendencia_ciclo']].tail(4)

# Convert time into numeric format for linear regression
last_4_trend['time_numeric'] = np.arange(len(last_4_trend))

# Step 4: Linear regression on the last 4 trend points
model = LinearRegression()
X = last_4_trend[['time_numeric']]
y = last_4_trend['emae_tendencia_ciclo']
model.fit(X, y)

# Step 5: Forecast the next 6 months using the linear model
future_months = np.arange(len(last_4_trend), len(last_4_trend) + 6).reshape(-1, 1)
trend_forecast = model.predict(future_months)

# Step 6: Calculate median of original - trend differences per month
data_2017_onwards['month'] = data_2017_onwards['indice_tiempo'].dt.month
data_2017_onwards['original_minus_trend'] = data_2017_onwards['emae_original'] - data_2017_onwards['emae_tendencia_ciclo']
monthly_medians = data_2017_onwards.groupby('month')['original_minus_trend'].median()

# Step 7: Forecast the next 6 months using trend + monthly medians
# Create a DataFrame to store future dates
last_date = data_2017_onwards['indice_tiempo'].max()
future_dates = pd.date_range(last_date + pd.DateOffset(months=1), periods=6, freq='MS')

# Match the forecast months with their corresponding median differences
future_medians = monthly_medians[future_dates.month.values].values

# Step 8: Final forecast: trend + median differences
emae_forecast = trend_forecast + future_medians

# Step 9: Plot the forecast along with historical data
plt.figure(figsize=(10, 6))

# Plot historical data
plt.plot(data_2017_onwards['indice_tiempo'], data_2017_onwards['emae_original'], label='Original', color='blue', linewidth=2)
plt.plot(data_2017_onwards['indice_tiempo'], data_2017_onwards['emae_tendencia_ciclo'], label='Trend', color='red', linestyle=':', linewidth=2)

# Plot forecast
plt.plot(future_dates, emae_forecast, label='Forecast', color='orange', linestyle='--', linewidth=2, marker = 's')

# Adding labels and title
plt.title('EMAE Decomposition and Forecast (Next 6 Months)')
plt.xlabel('Date')
plt.ylabel('EMAE Index')
plt.legend()
plt.grid(True)
plt.xticks(rotation=45)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Step 1: Load the dataset from the given URL
url = 'https://infra.datos.gob.ar/catalog/sspm/dataset/444/distribution/444.1/download/canastas-basicas-ciudad-de-buenos-aires.csv'
canasta_data = pd.read_csv(url)

# Step 2: Convert 'indice_tiempo' to datetime and filter data starting from 2023
canasta_data['indice_tiempo'] = pd.to_datetime(canasta_data['indice_tiempo'])
canasta_2023_onwards = canasta_data[canasta_data['indice_tiempo'] >= '2023-01-01']

# Step 3: Linear extrapolation for 'canasta_basica_alimentaria' and 'canasta_basica_total'
# Use the last 4 months of data for extrapolation
last_4_months = canasta_2023_onwards.tail(4)

# Prepare data for linear regression
X = np.arange(len(last_4_months)).reshape(-1, 1)  # Numeric indices for linear regression
y_alimentaria = last_4_months['canasta_basica_alimentaria']
y_total = last_4_months['canasta_basica_total']

# Fit linear regression models
model_alimentaria = LinearRegression().fit(X, y_alimentaria)
model_total = LinearRegression().fit(X, y_total)

# Step 4: Extrapolate until January 2025
# Generate future dates
last_date = canasta_2023_onwards['indice_tiempo'].max()
future_dates = pd.date_range(last_date + pd.DateOffset(months=1), periods=5, freq='MS')  # Up to Jan 2025

# Prepare future X values (numeric indices)
future_X = np.arange(len(last_4_months), len(last_4_months) + len(future_dates)).reshape(-1, 1)

# Make predictions using the linear models
alimentaria_forecast = model_alimentaria.predict(future_X)
total_forecast = model_total.predict(future_X)

# Step 5: Plot the data and extrapolated values
plt.figure(figsize=(10, 6))

# Plot historical Canasta Básica Alimentaria
plt.plot(canasta_2023_onwards['indice_tiempo'], canasta_2023_onwards['canasta_basica_alimentaria'], label='Canasta Básica Alimentaria (Historical)', color='blue', linewidth=2, marker = 'd')

# Plot extrapolated Canasta Básica Alimentaria
plt.plot(future_dates, alimentaria_forecast, label='Canasta Básica Alimentaria (Extrapolated)', color='blue', linestyle='--', linewidth=2, marker = 'd')

# Plot historical Canasta Básica Total
plt.plot(canasta_2023_onwards['indice_tiempo'], canasta_2023_onwards['canasta_basica_total'], label='Canasta Básica Total (Historical)', color='red', linewidth=2, marker = 'd')

# Plot extrapolated Canasta Básica Total
plt.plot(future_dates, total_forecast, label='Canasta Básica Total (Extrapolated)', color='red', linestyle='--', linewidth=2, marker = 'd')




# Step 1: Load the IPC and RIPTE datasets from their respective URLs
ipc_url = 'https://infra.datos.gob.ar/catalog/sspm/dataset/145/distribution/145.9/download/indice-precios-al-consumidor-apertura-por-categorias-base-diciembre-2016-mensual.csv'
ripte_url = 'https://infra.datos.gob.ar/catalog/sspm/dataset/158/distribution/158.1/download/remuneracion-imponible-promedio-trabajadores-estables-ripte-total-pais-pesos-serie-mensual.csv'

ipc_data = pd.read_csv(ipc_url)
ripte_data = pd.read_csv(ripte_url)

# Step 2: Convert 'indice_tiempo' to datetime and filter data starting from 2023
ipc_data['indice_tiempo'] = pd.to_datetime(ipc_data['indice_tiempo'])
ripte_data['indice_tiempo'] = pd.to_datetime(ripte_data['indice_tiempo'])

ipc_2023_onwards = ipc_data[ipc_data['indice_tiempo'] >= '2023-01-01']
ripte_2023_onwards = ripte_data[ripte_data['indice_tiempo'] >= '2023-01-01']

# Step 3: Linear extrapolation for IPC and RIPTE using the last 4 months
# Prepare data for linear regression for IPC
last_4_ipc = ipc_2023_onwards.tail(4)
X_ipc = np.arange(len(last_4_ipc)).reshape(-1, 1)
y_ipc = last_4_ipc['ipc_nivel_general_nacional']

# Fit the linear regression model for IPC
model_ipc = LinearRegression().fit(X_ipc, y_ipc)

# Prepare data for linear regression for RIPTE
last_4_ripte = ripte_2023_onwards.tail(4)
X_ripte = np.arange(len(last_4_ripte)).reshape(-1, 1)
y_ripte = last_4_ripte['ripte']

# Fit the linear regression model for RIPTE
model_ripte = LinearRegression().fit(X_ripte, y_ripte)

# Step 4: Extrapolate until January 2025 for both IPC and RIPTE
# Generate future dates
last_date_ipc = ipc_2023_onwards['indice_tiempo'].max()
last_date_ripte = ripte_2023_onwards['indice_tiempo'].max()

future_dates_ipc = pd.date_range(last_date_ipc + pd.DateOffset(months=1), periods=5, freq='MS')
future_dates_ripte = pd.date_range(last_date_ripte + pd.DateOffset(months=1), periods=6, freq='MS')

# Prepare future X values (numeric indices) for prediction
future_X_ipc = np.arange(len(last_4_ipc), len(last_4_ipc) + len(future_dates_ipc)).reshape(-1, 1)
future_X_ripte = np.arange(len(last_4_ripte), len(last_4_ripte) + len(future_dates_ripte)).reshape(-1, 1)

# Make predictions using the linear models
ipc_forecast = model_ipc.predict(future_X_ipc)
ripte_forecast = model_ripte.predict(future_X_ripte)

# # Step 5: Plot historical and extrapolated IPC and RIPTE
# plt.figure(figsize=(10, 6))

# Plot historical IPC
plt.plot(ipc_2023_onwards['indice_tiempo'], 100*ipc_2023_onwards['ipc_nivel_general_nacional'], label='IPC Nivel General Nacional (Historical)', color='.7', linewidth=2, marker='d')

# Plot extrapolated IPC
plt.plot(future_dates_ipc, 100*ipc_forecast, label='IPC Nivel General Nacional (Extrapolated)', color='.7', linestyle='--', linewidth=2, marker='d')

# Plot historical RIPTE
plt.plot(ripte_2023_onwards['indice_tiempo'], ripte_2023_onwards['ripte'], label='RIPTE (Historical)', color='orange', linewidth=2, marker='d')

# Plot extrapolated RIPTE
plt.plot(future_dates_ripte, ripte_forecast, label='RIPTE (Extrapolated)', color='orange', linestyle='--', linewidth=2, marker='d')




# Adding labels and title
plt.title('Canasta Básica Alimentaria y Total with Extrapolation to January 2025')
plt.xlabel('Date')
plt.ylabel('Value (in currency)')
plt.grid(True)
plt.xticks(rotation=45)
plt.legend()

# Display the plot
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# Initialize the number of months to extrapolate
# n_months = 20

# # Function to perform linear extrapolation
# def linear_extrapolation(df, column, n_months):
#     last_4_values = df[column].dropna().tail(4)
#     if len(last_4_values) < 4:
#         return pd.Series([np.nan] * n_months)  # Not enough data for extrapolation

#     # Linear model for extrapolation
#     X = np.arange(len(last_4_values)).reshape(-1, 1)
#     y = last_4_values.values
#     model = LinearRegression().fit(X, y)
    
#     # Generate future predictions
#     future_X = np.arange(len(last_4_values), len(last_4_values) + n_months).reshape(-1, 1)
#     forecast = model.predict(future_X)
    
#     return pd.Series(forecast)


def linear_extrapolation(df, column, n_months):
    # Extract the last 4 non-NaN values for the column
    last_4_values = df[column].dropna().tail(4)
    
    if len(last_4_values) < 4:
        return pd.Series([np.nan] * n_months)  # Not enough data for extrapolation

    # Linear model for extrapolation
    X = np.arange(len(last_4_values)).reshape(-1, 1)
    y = last_4_values.values
    model = LinearRegression().fit(X, y)
    
    # Generate future predictions for the next n_months
    future_X = np.arange(len(last_4_values), len(last_4_values) + n_months).reshape(-1, 1)
    forecast = model.predict(future_X)
    
    # Create a date range for the future months
    last_valid_date = df[column].dropna().index.max()  # Ensure the index is 'indice_tiempo'
    future_dates = pd.date_range(start=last_valid_date + pd.DateOffset(months=1), periods=n_months, freq='MS')
    
    # Return the forecast as a pandas Series with future_dates as index
    return pd.Series(forecast, index=future_dates)



# Function to perform trend + seasonal extrapolation
def trend_plus_seasonality_extrapolation(df, column_trend, column_original, n_months):
    # # Ensure 'indice_tiempo' is the index
    # if df.index.name != 'indice_tiempo':
    #     df = df.set_index('indice_tiempo')

    # Extract the last 4 non-NaN trend values for the column
    last_4_trend = df[[column_trend]].dropna().tail(4)

    if len(last_4_trend) < 4:
        # return pd.Series([np.nan] * n_months)  # Not enough data for extrapolation
        return pd.Series([np.nan] * n_months, index=pd.date_range(df.index.max() + pd.DateOffset(months=1), periods=n_months, freq='MS'))
        
    last_4_trend['time_numeric'] = np.arange(len(last_4_trend))
    
    # Fit linear trend model
    X_trend = last_4_trend[['time_numeric']]
    y_trend = last_4_trend[column_trend]
    trend_model = LinearRegression().fit(X_trend, y_trend)
    
    # Forecast trend component for the next months
    future_X_trend = np.arange(len(last_4_trend), len(last_4_trend) + n_months).reshape(-1, 1)
    trend_forecast = trend_model.predict(future_X_trend)
    
    # Step 6: Calculate monthly deviations from the trend
    df['month'] = df.index.month
        # we need to somehow do .loc[dfs time index values], so to not have problems.
    # Hint... we need to load the csv 
    # df['residuals'] = info[column_original] - info[column_trend]
    df['residuals'] = df[column_original] - df[column_trend]
    
    # Group by month and calculate median residuals
    monthly_residuals = df.groupby('month')['residuals'].median()
    
    # Get the future months for the forecast
    future_dates = pd.date_range(last_4_trend.index.max() + pd.DateOffset(months=1), periods=n_months, freq='MS')
    future_months = future_dates.month
    
    # Apply the median residuals to the trend forecast
    future_residuals = monthly_residuals[future_months].values
    final_forecast = trend_forecast + future_residuals
    
    return pd.Series(final_forecast, index=future_dates)

# # Example usage for EMAE
# n_months = 6
# emae_forecast = trend_plus_seasonality_extrapolation(df, column_trend='emae_tendencia_ciclo', column_original='emae_original', n_months=n_months)

# # Example usage for Employment series (asalariados_priv_estacionalidad)
# empleo_privado_forecast = trend_plus_seasonality_extrapolation(df, column_trend='asalariados_priv_estacionalidad', column_original='asalariados_priv_sin_estac', n_months=n_months)


def extrapolate_and_extend(df, column_trend=None, column_original=None, method=None, n_months=6):
    # # Ensure 'indice_tiempo' is the index
    # if df.index.name != 'indice_tiempo':
    #     df = df.set_index('indice_tiempo')
    
    # Extrapolation logic
    if method == 'linear':
        # Find the last valid date based on the column's non-null values
        last_valid_date = df[column_trend].dropna().index.max()
        future_dates = pd.date_range(start=last_valid_date + pd.DateOffset(months=1), periods=n_months, freq='MS')
        extrapolated_values = linear_extrapolation(df, column_trend, n_months)
        
    elif method == 'trend_plus_median':
        # Find the last valid date based on the trend column
        last_valid_date = df[column_original].dropna().index.max()
        future_dates = pd.date_range(start=last_valid_date + pd.DateOffset(months=1), periods=n_months, freq='MS')
        extrapolated_values = trend_plus_seasonality_extrapolation(df, column_trend=column_trend, column_original=column_original, n_months=n_months)
    
    else:
        raise ValueError("Unsupported method")
    
    # Create a DataFrame with the future dates and extrapolated values
    extrapolated_df = pd.DataFrame(extrapolated_values, index=future_dates, columns=[column_original if column_original else column_trend])

    # Extend the original DataFrame by appending the new extrapolated values
    df_extended = df.combine_first(extrapolated_df)
    
    return df_extended




# Apply extrapolation for each series individually
columns_to_extrapolate = {
    'Canasta_Basica_Total': 'linear',
    'RIPTE': 'linear',
    'IPC': 'linear',
    'EMAE': ('trend_plus_median', 'emae_tendencia_ciclo', 'emae_original'),
    'Empleo_Privado': ('trend_plus_median', 'asalariados_priv_estacionalidad', 'asalariados_priv_sin_estac'),
    'Empleo_Publico': ('trend_plus_median', 'asalariados_pub_estacionalidad', 'asalariados_pub_sin_estac'),
    'Empleo_Resto': ('trend_plus_median', 'resto_trab_estacionalidad', 'resto_trab_sin_estac')
}
# emae_data['seasonality_profile'] = emae_data['emae_original'] - emae_data['emae_tendencia_ciclo']

```
